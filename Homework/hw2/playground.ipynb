```python
b, q, d_model, n, h = 2, 5, 8, 2, 4  # d_model must equal n*h
x = torch.randn(b, q, d_model)
w_q = torch.randn(d_model, n, h)

q = torch.einsum("bqd,dnh->bqnh", x, w_q)
print(q.shape)  # torch.Size([2, 5, 2, 4])
```
